```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error , r2_score
```

#### Step 1: Data Collection and Import
#### Load the dataset from a URL or local file path

```python

data_url = "https://raw.githubusercontent.com/ageron/handson-ml2/master/datasets/housing/housing.csv"
df = pd.read_csv(data_url)
```

In the feature called "Total_bedroom" with the meadin of the column

```python
duplicates = df.duplicated().sum()
print("Duplicate rows:", duplicates)
```

```python
median_bedrooms = df["total_bedrooms"].median()
median_bedrooms
```

```python
df["total_bedrooms"].isnull().sum()
```

### Enhancing the Model: One-Hot Encoding and Feature Engineering

```python
df.fillna(median_bedrooms,inplace=True)
```

```python
df["total_bedrooms"].isnull().sum()
```

```python
df.info()
```

```python
# Apply one-hot encoding to 'ocean_proximity'
df_encoded = pd.get_dummies(df, columns=['ocean_proximity'])

df_encoded.head()
```

### Feature Engineering.. 

> Creating new features (based on other features (columns,inputs,vectors) )

```python
# Create new features
df_encoded['rooms_per_household'] = df_encoded['total_rooms'] / df_encoded['households']
df_encoded['bedrooms_per_room'] = df_encoded['total_bedrooms'] / df_encoded['total_rooms']
df_encoded['population_per_household'] = df_encoded['population'] / df_encoded['households']

df_encoded.head()
```

### Vertical split (to seprate input & output)


```python
x=df_encoded.drop(columns="median_house_value")
#input
```

```python
x.columns
# there is no median_house_value column now
```

```python
y=df_encoded['median_house_value']
#storing the input data as output data
```

```python
y
```

##  Horizontal Split (Train, Test) 

```python
from sklearn.model_selection import train_test_split
```

```python
# Split the data into training (80%) and testing (20%) sets
x_train_new, x_test_new, y_train_new, y_test_new = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
x_train_new
```

standardisation 

```python
from sklearn.preprocessing import StandardScaler
# Initalize the StandardScaler
scaler_new = StandardScaler()

# Fit & transform the training features
x_train_scaled_new = scaler_new.fit_transform(x_train_new) # mean & standard deveation

# Transform the testing features using the training set scaler
x_test_scaled_new = scaler_new.transform(x_test_new) #mean & standard deveation
```

```python
## use the regression model
from sklearn.linear_model import LinearRegression

lr=LinearRegression()
```

```python
pd.DataFrame(x_test_scaled_new,y_test_new).info()
```

```python
lr.fit(x_train_scaled_new,y_train_new)
```

```python
y_predict = lr.predict(x_test_scaled_new)
```

```python
## model evaluation 
from sklearn.metrics import r2_score,root_mean_squared_error
print("RMSE:",root_mean_squared_error(y_test_new,y_predict))
print("R2_Score:",r2_score(y_test_new,y_predict))
```
